```python
import pandas as pd
from sklearn.naive_bayes import CategoricalNB
from sklearn.preprocessing import OrdinalEncoder

# create dataframe
data = pd.DataFrame({
    'Outlook': ['Sunny','Sunny','Overcast','Rain','Rain','Sunny'],
    'Temperature': ['Hot','Hot','Hot','Mild','Cool','Mild'],
    'Humidity': ['High','High','High','High','Normal','Normal'],
    'Wind': ['Weak','Strong','Weak','Weak','Weak','Strong'],
    'PlayTennis': ['No','No','Yes','Yes','Yes','Yes']
})
```

```python
# split
X = data.drop('PlayTennis', axis=1)
y = data['PlayTennis']
```

```python

# encode categorical data

encoder = OrdinalEncoder()
X_encoded = encoder.fit_transform(X)

# train model
model = CategoricalNB()
model.fit(X_encoded, y)

# predict example
sample = encoder.transform([['Rain','Mild','Normal','Strong']])
print(model.predict(sample))
```

```python
import pickle

def save_model(model, filename):
    with open(filename, 'wb') as file:
        pickle.dump(model, file)
```

```python
save_model(model, 'Naive_Bayes_model.pkl')
```
