```python
X = df.iloc[:,0:8]
```

```python
y = df.iloc[:,-1]
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc = StandardScaler()
```

```python
X = sc.fit_transform(X)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2)
```

```python
len(X_test)
```

```python
len(y_train)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dt = DecisionTreeClassifier().fit(X_train,y_train)
```

```python
y_predict = dt.predict(X_test)
```

```python
# 2. Method A: Direct score (Accuracy)
print("Accuracy:", dt.score(X_test, y_test))
```

```python
from sklearn.ensemble import BaggingClassifier
```

```python
beg = BaggingClassifier(estimator=DecisionTreeClassifier(),oob_score=False, n_estimators=500, random_state=0)
```

```python
bag_clf = BaggingClassifier( DecisionTreeClassifier(), n_estimators=500,oob_score=False, max_samples=100, bootstrap=True, n_jobs=-1)
```

```python
beg.fit(X_train,y_train)
```

```python
print("Accuracy of Beg:", beg.score(X_test, y_test))
```

```python
bag_clf.fit(X_train,y_train)
```

```python
print("Accuracy of Beg_clf:", bag_clf.score(X_test, y_test))
```
