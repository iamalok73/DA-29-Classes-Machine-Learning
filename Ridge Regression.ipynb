```python
X = df.iloc[:,0:4].values
y = df.iloc[:,-1].values
```

```python
# dummy_encoder
# ordinal Encoding
# one hot encoding
# Column Transformation
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
from sklearn.compose import ColumnTransformer
```

```python
ct = ColumnTransformer(
    transformers=[
       ("tf1", OneHotEncoder(drop="first", sparse_output=False), [3])
    ],
    remainder="passthrough"
                      )
```

```python
X = np.array(ct.fit_transform(X))
```

```python
X
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc = StandardScaler()
```

```python
X = sc.fit_transform(X)
```

```python
X
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2)
```

```python
len(y_train)
```

```python
from sklearn.linear_model import LinearRegression
```

```python

len(y_train)
```

```python
model = LinearRegression()
```

```python
model.fit(X_train,y_train)
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
accuracy = model.score(X_test, y_test)
print("Accuracy:", round(accuracy*100,2))
print("r2 Score: ",r2_score(y_test,y_pred))
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(y_pred)
plt.plot(y_test)
```

```python
from sklearn.linear_model import Ridge,Lasso
```

```python
ridge = Ridge(alpha=0.3)
ridge.fit(X_train, y_train)
pred_basic = ridge.predict(X_test)
print(pred_basic)
```

```python
accuracy = ridge.score(X_test, y_test)
print("Accuracy:", round(accuracy*100,2))
```

```python
print("MSE (alpha = 1.0):", mean_squared_error(y_test, pred_basic))
```

```python
y_pred = ridge.predict(X_test)

mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

mse, r2
```

```python
print("Linear model coffecient : ",model.coef_)
```

```python
print("Coefficients (alpha = 1.0):", ridge.coef_)
```

```python
lasso = Lasso(alpha=0.1)
lasso.fit(X_train, y_train)
```

```python
y_pred = lasso.predict(X_test)

mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

mse, r2,y_pred
```

```python
y_pred = lasso.predict(X_test)
print(y_pred)
plt.figure(figsize=(6, 6))
plt.scatter(y_test, y_pred)

min_val = min(y_test.min(), y_pred.min())
max_val = max(y_test.max(), y_pred.max())
plt.plot([min_val, max_val], [min_val, max_val])

plt.xlabel("Actual Values")
plt.ylabel("Predicted Values")
plt.title("Actual vs Predicted (Lasso Regression)")
plt.show()
```
