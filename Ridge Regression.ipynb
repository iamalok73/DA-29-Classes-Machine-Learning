```python
X = df.iloc[:,0:4].values
y = df.iloc[:,-1].values
```

```python
# dummy_encoder
# ordinal Encoding
# one hot encoding
# Column Transformation
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
from sklearn.compose import ColumnTransformer
```

```python
ct = ColumnTransformer(
    transformers=[
       ("tf1", OneHotEncoder(drop="first", sparse_output=False), [3])
    ],
    remainder="passthrough"
                      )
```

```python
X = np.array(ct.fit_transform(X))
```

```python
X
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc = StandardScaler()
```

```python
X = sc.fit_transform(X)
```

```python
X
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2)
```

```python
from sklearn.linear_model import LinearRegression
```

```python

len(y_train)
```

```python
model = LinearRegression()
```

```python
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(y_pred)
plt.plot(y_test)
```

```python
accuracy = model.score(X_test, y_test)
print("Accuracy:", round(accuracy*100,2))
```

```python
from sklearn.linear_model import Ridge
```

```python
ridge = Ridge(alpha=4)
ridge.fit(X_train, y_train)
pred_basic = ridge.predict(X_test)
print(pred_basic)
```

```python
accuracy = ridge.score(X_test, y_test)
print("Accuracy:", round(accuracy*100,2))
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error
```

```python
print("MSE (alpha = 1.0):", mean_squared_error(y_test, pred_basic))
```

```python
print("Linear model coffecient : ",model.coef_)
```

```python
print("Coefficients (alpha = 1.0):", ridge.coef_)
```
