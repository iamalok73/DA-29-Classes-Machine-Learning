```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.feature_selection import RFE
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv(r"C:\Users\AlokVerma\Desktop\ML DATA\Telco_Cusomer_Churn.csv")
```

```python
df.head()
```

```python
df.drop(columns=["customerID"],axis="columns", inplace=True)
```

```python
df["TotalCharges"]= pd.to_numeric(df["TotalCharges"],errors="coerce")
```

```python
df.dropna(inplace=True,how="any")
```

```python
df = pd.get_dummies(df,drop_first=True)
```

```python
df.Churn_Yes.value_counts()
```

```python
X = df.drop(["Churn_Yes"],axis=1)
y = df["Churn_Yes"]
```

```python
X
```

```python
y.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
model_orig = LogisticRegression()
```

```python
model_orig.fit(X_train,y_train)
```

```python
pred_orig = model_orig.predict(X_test)
```

```python
accuracy_orig = accuracy_score(y_test,pred_orig)
```

```python
accuracy_orig
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dt  = DecisionTreeClassifier()
```

```python
dt.fit(X_train,y_train)
```

```python
dt_pred = dt.predict(X_test)
```

```python
accuracy_score(y_test,dt_pred)
```

# Applying PCA Algorithms

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components = 4)
```

```python
X_train_pcs = pca.fit_transform(X_train)
X_test_pca = pca.fit_transform(X_test)
```

```python
explained_variance = pca.explained_variance_

print(explained_variance)
```

```python
# Training the logistics Regeression
```

```python
model_log = LogisticRegression()
```

```python
model_log.fit(X_train_pcs,y_train)
```

```python
pca_pred = model_log.predict(X_test_pca)
```

```python
pca_accuracy = accuracy_score(y_test,pca_pred)
```

```python
pca_accuracy
```

```python
dt_pca = DecisionTreeClassifier()
```

```python
dt_pca.fit(X_train_pcs,y_train)
```

```python
dt_pca_pred = dt_pca.predict(X_test_pca)
```

```python
accuracy_score(y_test,dt_pca_pred)
```
